#updating requirements file, for some reason grok says to do this as a workaround for mk library and numpy being out of synch.

```python
# Cell 1: FORCE remove mkl + conda-pack (guaranteed to work)
import os

req_file = '/content/Vaccine/vaccine_pip.txt'

with open(req_file, 'r') as f:
    lines = f.readlines()

with open(req_file, 'w') as f:
    commented = 0
    for line in lines:
        stripped = line.strip()
        if stripped.startswith(('mkl-fft==', 'mkl-service==', 'mkl-random', 'conda-pack==')):
            f.write(f'#{stripped}   # commented - no conda / Colab conflict\n')
            commented += 1
        else:
            f.write(line)
    print(f"✅ Successfully commented out {commented} conflicting packages")
    print("   (mkl-fft, mkl-service, mkl-random, conda-pack)")

# Show proof
print("\n🔍 Remaining mkl/conda lines in file:")
!grep -E '^(#.*(mkl|conda-pack)|mkl|conda-pack)' {req_file} || echo "   (none left - perfect)"
```
